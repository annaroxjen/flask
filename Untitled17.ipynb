### 5. Compare the Performance of Different Models
Let's extract the evaluated metrics (`Accuracy` and `ROC AUC`) from the trained models to easily identify the best performing classifier.

```python
import pandas as pd

# Compile the results of the trained models into a single DataFrame
model_comparison = []
for model_name, metrics in results.items():
    model_comparison.append({
        'Model': model_name,
        'Accuracy': metrics['Accuracy'],
        'ROC AUC': metrics['ROC AUC']
    })

compare_df = pd.DataFrame(model_comparison).sort_values(by='ROC AUC', ascending=False)
display(compare_df)
```

### Save the Best Model
We will save the best-performing model pipeline (the preprocessing steps and classifier integrated together) as a serialized pickle file so it can be loaded directly inside our Flask web application.

```python
import pickle

best_model_name = compare_df.iloc[0]['Model']
best_pipeline = results[best_model_name]['Pipeline']

print(f"Saving the best model: {best_model_name}")
with open('best_model.pkl', 'wb') as f:
    pickle.dump(best_pipeline, f)
```

### 6. Create a Website using Flask with Necessary Features
We will now write a python script `app.py` that serves as our web application backend. It includes an interactive user interface using a Bootstrap form to receive patient data and output live risk predictions using our saved model.

```python
flask_app_code = """
import pickle
import pandas as pd
from flask import Flask, request, render_template_string

app = Flask(__name__)

# Load the pre-trained pipeline model
with open('best_model.pkl', 'rb') as f:
    model = pickle.load(f)

HTML_TEMPLATE = '''
<!DOCTYPE html>
<html>
<head>
    <title>Diabetes Prediction System</title>
    <link rel="stylesheet" href="https://cdn.jsdelivr.net/npm/bootstrap@5.3.0/dist/css/bootstrap.min.css">
</head>
<body class="bg-light">
    <div class="container py-5">
        <div class="row justify-content-center">
            <div class="col-md-8">
                <div class="card shadow">
                    <div class="card-header bg-primary text-white">
                        <h3 class="text-center mb-0">Diabetes Risk Predictor</h3>
                    </div>
                    <div class="card-body">
                        <form method="POST" action="/predict">
                            <div class="row">
                                <div class="col-md-6 mb-3">
                                    <label class="form-label">Gender</label>
                                    <select name="gender" class="form-select" required>
                                        <option value="Female">Female</option>
                                        <option value="Male">Male</option>
                                        <option value="Other">Other</option>
                                    </select>
                                </div>
                                <div class="col-md-6 mb-3">
                                    <label class="form-label">Age</label>
                                    <input type="number" name="age" class="form-control" step="1" min="0" max="120" required>
                                </div>
                            </div>

                            <div class="row">
                                <div class="col-md-6 mb-3">
                                    <label class="form-label">Hypertension</label>
                                    <select name="hypertension" class="form-select" required>
                                        <option value="0">No</option>
                                        <option value="1">Yes</option>
                                    </select>
                                </div>
                                <div class="col-md-6 mb-3">
                                    <label class="form-label">Heart Disease</label>
                                    <select name="heart_disease" class="form-select" required>
                                        <option value="0">No</option>
                                        <option value="1">Yes</option>
                                    </select>
                                </div>
                            </div>

                            <div class="row">
                                <div class="col-md-6 mb-3">
                                    <label class="form-label">Smoking History</label>
                                    <select name="smoking_history" class="form-select" required>
                                        <option value="never">Never</option>
                                        <option value="No Info">No Info</option>
                                        <option value="current">Current</option>
                                        <option value="former">Former</option>
                                        <option value="not current">Not Current</option>
                                        <option value="ever">Ever</option>
                                    </select>
                                </div>
                                <div class="col-md-6 mb-3">
                                    <label class="form-label">BMI</label>
                                    <input type="number" name="bmi" class="form-control" step="0.01" min="5" max="100" required>
                                </div>
                            </div>

                            <div class="row">
                                <div class="col-md-6 mb-3">
                                    <label class="form-label">HbA1c Level</label>
                                    <input type="number" name="HbA1c_level" class="form-control" step="0.1" min="3" max="15" required>
                                </div>
                                <div class="col-md-6 mb-3">
                                    <label class="form-label">Blood Glucose Level</label>
                                    <input type="number" name="blood_glucose_level" class="form-control" step="1" min="50" max="400" required>
                                </div>
                            </div>

                            <button type="submit" class="btn btn-primary w-100">Predict</button>
                        </form>
                    </div>
                </div>

                {% if prediction_text %}
                <div class="card mt-4 shadow">
                    <div class="card-body text-center">
                        <h4>Result:</h4>
                        <div class="alert {% if result_class == 'danger' %}alert-danger{% else %}alert-success{% endif %}">
                            {{ prediction_text }}
                        </div>
                    </div>
                </div>
                {% endif %}
            </div>
        </div>
    </div>
</body>
</html>
'''

@app.route('/', methods=['GET'])
def home():
    return render_template_string(HTML_TEMPLATE)

@app.route('/predict', methods=['POST'])
def predict():
    # Retrieve values from web page form submission
    input_data = {
        'gender': request.form['gender'],
        'age': float(request.form['age']),
        'hypertension': int(request.form['hypertension']),
        'heart_disease': int(request.form['heart_disease']),
        'smoking_history': request.form['smoking_history'],
        'bmi': float(request.form['bmi']),
        'HbA1c_level': float(request.form['HbA1c_level']),
        'blood_glucose_level': float(request.form['blood_glucose_level'])
    }

    # Convert inputs to DataFrame matching input specifications
    df_input = pd.DataFrame([input_data])

    # Generate predictions and prediction probabilities
    prediction = model.predict(df_input)[0]
    probability = model.predict_proba(df_input)[0][1]

    if prediction == 1:
        prediction_text = f"The patient is high risk of diabetes. (Confidence: {probability:.2%})"
        result_class = "danger"
    else:
        prediction_text = f"The patient is low risk of diabetes. (Confidence: {(1-probability):.2%})"
        result_class = "success"

    return render_template_string(HTML_TEMPLATE, prediction_text=prediction_text, result_class=result_class)

if __name__ == '__main__':
    app.run(host='0.0.0.0', port=5000)
"""

with open('app.py', 'w') as f:
    f.write(flask_app_code)
print("Flask app file 'app.py' successfully saved!")
```

1. Create a Python program to generate a Fibonacci sequence up to a specified
number

```python
def generate_fibonacci_up_to(limit):

    if limit < 0:
        return []

    sequence = [0]
    if limit == 0:
        return sequence

    a, b = 0, 1
    while b <= limit:
        sequence.append(b)
        a, b = b, a + b

    return sequence


if __name__ == "__main__":
    max_value = 100
    fib_sequence = generate_fibonacci_up_to(max_value)
    print(f"Fibonacci sequence up to {max_value}:")
    print(fib_sequence)
```

2. Write a function to rotate a given list by a specified number of positions.

```python
def rotate_list(lst, k)
    if not lst:
        return lst


    k = k % len(lst)


    return lst[-k:] + lst[:-k]

if __name__ == "__main__":
    my_list = [1, 2, 3, 4, 5]


    print(rotate_list(my_list, 2))   # Output: [4, 5, 1, 2, 3]


    print(rotate_list(my_list, -1))  # Output: [2, 3, 4, 5, 1]
```
